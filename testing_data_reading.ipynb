```python
import pandas as pd
```

```python
df = pd.read_parquet('grade_summary.parquet')
```

```python
df.columns
```

```python
class StudentBase():
    id : int 
    name : str | None = None
    age : int
    social_data : int | None = None
    overall_grade : str
    fundamental_comp : str
    behavioral_comp : str
    disabilities : str | None
```

```python
student_base = df[
    [
        'codigo', #id
        'nombre', #name
        #age,
        #social_data,
        'band', #overall
        'fundamental', #fundamental
        'compo'
    ]
].reset_index().drop(columns=['index'])
```

```python
student_base.codigo.value_counts().reset_index()[student_base.codigo.value_counts().reset_index()['count'] > 2]
```

```python
student_base['age'] = 16
```

```python
student_base['social_data'] = None
```

```python
student_base = student_base[['codigo', 'nombre', 'age', 'social_data', 'fundamental', 'band', 'compo']]
```

```python
student_base.rename(
    columns={
        "codigo" : "id",
        "nombre" : "name",
        "fundamental" : "fundamental_comp",
        "band" : "overall_grade",
        "compo" : "behavioral_comp"
    },
    inplace=True
)
```

```python
student_base["disabilities"] = None
```

```python
student_base
```

```python
student_base.to_json('student_info.json', orient='records')
```
